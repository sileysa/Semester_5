1. PyPREP

PyPREP adalah library Python yang digunakan untuk melakukan preprocessing pada data EEG (Electroencephalography). PyPREP merupakan implementasi dari PREP Pipeline dan digunakan untuk membantu membersihkan serta mempersiapkan data EEG sebelum dilakukan analisis lebih lanjut.

2. SciPy

SciPy adalah library Python yang digunakan untuk komputasi ilmiah dan matematika. SciPy menyediakan berbagai fungsi untuk statistik, optimasi, interpolasi, integrasi, aljabar linear, pemrosesan sinyal, serta berbagai perhitungan ilmiah lainnya.

3. wandb

wandb atau Weights & Biases merupakan library yang digunakan untuk melakukan experiment tracking pada machine learning. Library ini dapat digunakan untuk mencatat parameter, hyperparameter, nilai loss, accuracy, hasil training, serta menyimpan dan membandingkan hasil dari berbagai eksperimen.

4. pyECG

pyECG merupakan library Python yang digunakan untuk membaca dan menulis data ECG (Electrocardiogram). Library ini mendukung beberapa format dataset ECG, seperti format PhysioBank WFDB dan iShine Holter. Dengan demikian, pyECG dapat membantu dalam pengolahan data sinyal ECG untuk kebutuhan penelitian atau analisis.
